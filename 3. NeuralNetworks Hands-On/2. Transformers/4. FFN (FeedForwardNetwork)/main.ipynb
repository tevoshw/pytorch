```python
import torch
import torch.nn as nn


class FeedForwardNetwork(nn.Module):
    def __init__(self, d_model):
        super().__init__()


        self.ffn = nn.Sequential(
            nn.Linear(d_model, d_model * 4),
            nn.ReLU(),
            nn.Linear(d_model * 4, d_model)
        )


    def forward(self, x):

        x = self.ffn(x)
```
