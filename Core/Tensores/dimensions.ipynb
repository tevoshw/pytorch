```python
import torch

matrix1 = torch.rand(12)
matrix2 = torch.rand(16)
matrix3 = torch.rand(3, 4, 5)
matrix4 = torch.rand(3, 4)
matrix5 = torch.rand(3, 1, 4, 5)
```

# 1. Reshape & View

## 1. Reshape
Returns a tensor with the same data and number of elements as input, but with the specified shape. When possible, the returned tensor will be a view of input. Otherwise, it will be a copy. Contiguous inputs and inputs with compatible strides can be reshaped without copying, but you should not depend on the copying vs. viewing behavior

`tensorX.reshape(new dimensions)` Reshape the tensor dimensions

## 2. View

Returns a tensor with the same data and number of elements as input, but with the specified shape. The returned tensor is always a view of the original tensor (no data copy). The input tensor must be contiguous in memory; otherwise, the operation will fail

`tensorX.view(new dimensions)`

**The main difference between .view() and .reshape() in PyTorch lies in memory management and operational flexibility. Both change the shape (shape) of your tensor without changing the data, but they do it in different ways under the hood. While reshape is safer, view is faster**

```python
matrix1_reshape = matrix1.reshape(3, 4)
matrix2_reshape = matrix2.reshape(4, -1) 

matrix1_view = matrix1.view(3, 4)
matrix2_view = matrix2.view(4, -1) 

print(f'Matrix Before: {matrix1.shape}')
print(f'Matrix after reshape 1 (3, 4): {matrix1_reshape.shape}')
print(f'Matrix after view 1 (3, 4): {matrix1_view.shape}')

print(f'='* 50)

print(f'Matrix 2 before: {matrix2.shape}')
print(f'Matrix after reshape 2 (4, -1): {matrix2_reshape.shape}')
print(f'Matrix after view 2 (4, -1): {matrix2_view.shape}')

```

# 2. Tranpose & Permute


## 1. Tranpose
Inverts the dimensions only 2

`tensorX.tranpose(new dimensions)`

`.T` 

## 2. Permute

Inverters how much dimensions do you want

`tensorX.permute(new dimensions)`



```python
matrix3_tranpose = matrix3.transpose(1, 2)
matrix4_tranposeT = matrix4.T

print(f'Matrix 3 before: {matrix3.shape}')
print(f'Matrix after the tranpose (1, 2): {matrix3_tranpose.shape}')
print(f'='* 50)

print(f'Matrix 4 before: {matrix4.shape}')
print(f'Matrix 4 after the transpose (.T) : {matrix4_tranposeT.shape}')
```

```python
matrix3_permute = matrix3.permute(1, 2, 0)

print(f'Matrix 3 before: {matrix3.shape}')
print(f'Matrix after the permute (1, 2, 0): {matrix3_permute.shape}')
```

# 3. Squeeze & Unsqueeze

## 1. Unsqueeze
The unsqueeze adds a new dimension of size 1 at a specific position.

`tensorX.unsqueeze(dimension)`

## 2. Squeeze
Squeeze removes all size 1 dimensions from your tensor. If you pass a specific dimension index, it removes only that dimension as long as it has size 1.

`tensorX.squeeze(dimension)`



```python
matrix1_unsqueeze = matrix1.unsqueeze(1)
matrix5_squeeze = matrix5.squeeze(1)

print(f'Matrix 1 Before: {matrix1.shape}')
print(f'Matrix 1 after the unsqueeze (dim 1): {matrix1_unsqueeze.shape}')

print(f'='* 50)


print(f'Matrix 5 Before: {matrix5.shape}')
print(f'Matrix 5 after the squeeze (dim1): {matrix5_squeeze.shape}')
```

# 4. Flatten

It is the operation you use when you want to transform any multidimensional tensor (such as an image or a text string) into a one-dimensional (1D) vector or a set of flat vectors

`tensorX.flatten()`

dim0 * dim1 * dim2 * dimN

```python
matrix5_flatten = matrix5.flatten()

print(f'Matrix 5 Before: {matrix5.shape}')
print(f'Matrix 5 after the squeeze (dim1): {matrix5_flatten.shape}')
```
