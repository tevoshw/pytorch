# Weight Tying

Technique of 2 layers or something have the same weights

The `.weight` can acess the values of the weights

> We can acess and do a lot of things with that

```python
import torch.nn as nn

layer1 = nn.Linear(4, 4)
layer2 = nn.Linear(4, 4)

print(f'=' * 50)
print(f'Before the tying')
print(f'L1: {layer1.weight}')
print(f'L2: {layer2.weight}')

layer2.weight = layer1.weight

print(f'=' * 50)
print(f'After the tying')
print(f'L1: {layer1.weight}')
print(f'L2: {layer2.weight}')

```

# Others cool things

```python
print(f'Shape of layer 1: {layer1.weight.shape}')
print(f'Acess the bias of layer1: {layer1.bias} and their shape: {layer1.bias.shape}')
print(f'Acess the bias of layer2: {layer2.bias}')
```
