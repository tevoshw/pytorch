# Apply 

the `.apply()` in the pytorch are a method inside the `nn.Module`that loops through all modules and their submodules. And the argument it's the name module

Connect to the /4. KERNEL INICIALIZATION

```python
import torch.nn as nn

class Test(nn.Module):
    def __init__(self):
        self.a = nn.Linear(2, 2)
        self.b = nn.Linear(2, 2)
        self.c = nn.Linear(2, 2)
        self.apply(self._init_weights) # Will pass by each module and submodule of the Model, so will pass the Name of the layer as argument

    def _init_weights(self, module): # So here we need to receive the module, and the module will be the layers name 
        if isistance(module, nn.Linear): # Check ifs the module are nn.Linear()
            nn.init_normal_(module.weight, mean=0.0, std=0.02) # the init_normal_ the _ in the final it's inplace
            if module.bias is not None:
                nn.init_zeros(module.bias)
        elif isistance(module, nn.Embedding): # Check ifs the module are embedding
            nn.init_normal_(module.weight, mean=0.0, std=0.02)
```
