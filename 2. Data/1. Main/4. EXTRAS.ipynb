# 1. Collate FN (Preparing batches I)

The collate is the function that turns a list of loose samples into a batch ready for the model. The DataLoader takes B samples from the dataset, each with its own size and format, and hands that list to the collate_fn, which is called once per batch.

We use it because a batch must be a rectangular tensor, and samples rarely arrive that way: sequences have different lengths, fields have different types, and sometimes data is missing or corrupted. Inside the collate, we split the fields (x and y), cut what exceeds the limit (truncation), fill in what is missing (padding), build the mask that separates real tokens from filler, and convert everything to tensors.

See more in: https://github.com/tevoshw/preparing_batches_ml/blob/main/main.ipynb

# 2. Packing (Preparing batches II)

Packing is a technique that concatenates multiple documents to fill fixed-length sequences, avoiding padding waste.

To prepare batches for NLP tasks (pre-training, SFT, etc.), it's worth studying how data is loaded, packed and padded.

See more in: https://github.com/tevoshw/packing_nlp
