# 1. Conv

### 1. Understand the data and their Conv TYPE

Basically in CNNs we receive as data

- Photos = Have 3 dimensions (CHANNELS, HEIGHT, WIDTH)
- Videos = Have 4 dimensions (CHANNELS, FRAME, HEIGHT, WIDTH)
- Audios = Have 2 dimensions (CHANNELS, TIME)


Our kernel (matriz of filters) will slide in some dimensions, in pytorch we gonna use the `ConvXd`, the *X* are the numbers of dimensions that our data have for the kernel slide

1. **Raw Audios (C, T)** = T (`Conv1d`)
2. **Photos (B, C, H, W)** = H and W (`Conv2d`)
3. **Videos (B, C, D, H, W)** = D, H and W (`Conv3d`)

---
### 2. Conv ARGS
- All the CONVS in PyTorch have the same args, like the example

```python

import torch.nn as nn

conv = nn.Conv2d(
    in_channels = 1, # The number of input channels (C_IN)
    out_channels = 10, # The number of output channels (C_OUT)
    kernel_size = INT for (INT, INT) or (X, y), # The H and W in shape of an matriz
    padding = 2,  # The number of padding
    stride = 3,  # The number of stride
    bias = False # If bias or not
)


```

---
### 3. Kernel Outputs (Feature Maps)

When we're building some model, it's really important know the shape while inside the model, so let's analize
> The main rule of this = (dimension - kDimensions + 2 * padding) / stride + 1

Example:

```
Input = (b, x, dim)
Kernel = (c_out, c_in, kDIM)
Output = (b, c_out, DIM')

dim' = (dim - k + 2*padding) / stride + 1

```


1. `Conv1d`:
- Input Data (B, X, Y)
- Kernel (C_out, cin, kY)
- Output = (B, C_out, Y')

2. `Conv2d`
- Input Data (B, X, Y, Z)
- Kernel (C_out, cin, kY, kZ)
- Output = (B, C_out, Y', Z')


3. `Conv3d`
- Input Data (B, W, X, Y, Z)
- Kernel (C_out, cin, kX, kY, kZ)
- Output = (B, C_out, X', Y', Z')

# 2. Pooling


# 1. Understand the dimensions

In PyTorch Pooling have some types like CONVs

1. `MaxPool1d()`
2. `MaxPool2d()`
3. `MaxPool3d()`

The rule here it's the same as CONVs, 'how many dimensions the kernel slide', for better understand just use the same XD as the CONV


# 2. Types of POOLING


The most common poolings are

1. MaxPooling, that choose the higher number of the window, for this we use the `MaxPoolxD()`
2. AveragePooling, that do a media for the window context, for this we use the `AvgPoolxd()`
3. GlobalAvgPooling, that do a media of the WHOLE feature map (not just a window), for this we use `AdaptiveAvgPoolxD(output_size=1)`
4. GlobalMaxPooling, that choose the highest number of the WHOLE feature map, for this we use `AdaptiveMaxPoolxD(output_size=1)`


# 3. Pool Args


```python

import torch.nn as nn


max_pool = nn.MaxPool2d(
    kernel_size = 2,
    stride = None, # Same as the kernel
    padding = 0,
    ceil_mode = True # If FLOATS round to higher number
)


avg_pool = nn.AvgPool2d(
    kernel_size = 2,
    stride = None, # Same as the kernel
    count_include_pad = True # If TRUE 0 of the padding enters in the count 
)


global_avg_pool = nn.AdaptiveAvgPool2d(
    output_size = 1 # Always reduces to 1x1, no matter the input size
)


global_max_pool = nn.AdaptiveMaxPool2d(
    output_size = 1 # Always reduces to 1x1, no matter the input size
)

```
