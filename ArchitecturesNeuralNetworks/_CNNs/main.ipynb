# 1. Conv

### 1. Understand the data and their Conv TYPE

Basically in CNNs we receive as data

- Photos = Have 3 dimensions (CHANNELS, HEIGHT, WIDTH)
- Videos = Have 4 dimensions (CHANNELS, FRAME, HEIGHT, WIDTH)
- Audios = Have 2 dimensions (CHANNELS, TIME)


Our kernel (matriz of filters) will slide in some dimensions, in pytorch we gonna use the `ConvXd`, the *X* are the numbers of dimensions that our data have for the kernel slide

1. **Raw Audios (C, T)** = T (`Conv1d`)
2. **Photos (B, C, H, W)** = H and W (`Conv2d`)
3. **Videos (B, C, D, H, W)** = D, H and W (`Conv3d`)

### 3. Kernel Outputs (Feature Maps)

When we're building some model, it's really important know the shape while inside the model, so let's analize
> The main rule of this = (dimension - kDimensions + 2 * padding) / stride + 1

Example:

```
Input = (b, x, dim)
Kernel = (c_out, c_in, kDIM)
Output = (b, c_out, DIM')

dim' = (dim - k + 2*padding) / stride + 1

```


1. `Conv1d`:
- Input Data (B, X, Y)
- Kernel (C_out, cin, kY)
- Output = (B, C_out, Y')

2. `Conv2d`
- Input Data (B, X, Y, Z)
- Kernel (C_out, cin, kY, kZ)
- Output = (B, C_out, Y', Z')


3. `Conv3d`
- Input Data (B, W, X, Y, Z)
- Kernel (C_out, cin, kX, kY, kZ)
- Output = (B, C_out, X', Y', Z')
