```python
import torch
import torch.nn as nn
import torch.nn.functional as F

x = torch.randn(4, 4)
attn_out = torch.randn(4, 4)
```

# 1. Add & Norm 1

```python
# LayerNorm1
ln1 = nn.LayerNorm(4)

# Add & Norm
x = x + attn_out
x = ln1(x)

print(x.shape)
```

# 2. FFN

```python
ffn = nn.Sequential(
    nn.Linear(4, 16),
    nn.GELU(),
    nn.Linear(16, 4)
)

ffn_out = ffn(x)
print(ffn_out.shape)
```

# 3. Add & Norm 2

```python
# LayerNorm2
ln2 = nn.LayerNorm(4)

# Add & Norm
x = x + ffn_out
x = ln2(x)

print(x.shape)
```
