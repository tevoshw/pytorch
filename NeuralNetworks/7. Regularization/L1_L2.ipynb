# Regularização no PyTorch (L1 & L2)

Regularização pode entrar em 3 pontos do loop de treino:

```
forward → loss → backward → [gradient] → step (optim)
             ↑                   ↑              ↑
            loss              gradient        optim
```

## 1. Loss — manual

Qualquer coisa somada à loss entra no grafo computacional. O `.backward()` diferencia automaticamente em relação a ela.

```python
# L2
l2 = sum(p.pow(2).sum() for p in model.parameters())
loss = criterion(output, target) + lambda_ * l2
loss.backward()

# L1
l1 = sum(p.abs().sum() for p in model.parameters())
loss = criterion(output, target) + lambda_ * l1
loss.backward()
```
Automaticamente fazendo isso adicinará no gradient pelo `.backward()`

## 2. Otimizador — nativo 

Weight decay aplicado diretamente no peso durante o `.step()`, fora do fluxo do gradiente:

```python
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=0.01)
```

Aqui temos várias opções, e para utilizar use no seu optimizador, normalmente o nome vem com `weight_decay `
