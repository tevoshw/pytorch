# 1. Neurons (`nn.Linear`)
 
`nn.Linear(in_features, out_features, bias = True)` is the PyTorch implementation of a single dense layer — a group of neurons operating in parallel over the same input.
 
Internally, it does exactly one thing:
 
```python
output = input @ weight.T + bias
```
 
That's it. This is the dot product $\mathbf{w}^\top \mathbf{x} + b$ from the theory, written in matrix form to handle multiple neurons at once.
 
 
---

````python
import torch
import torch.nn as nn


w1 = nn.Linear(in_features=10, out_features=40, bias = True)
````

# 2. Layers (`nn.Sequential`)
 
`nn.Sequential` is a container that stacks layers in order. When you call it with an input, it passes the data through each layer one by one — the output of each layer becomes the input of the next.
 
```python
model = nn.Sequential(
    nn.Linear(784, 128),
    nn.ReLU(),
    nn.Linear(128, 64),
    nn.ReLU(),
    nn.Linear(64, 10)
)
 
output = model(x)  # x flows through all 5 layers automatically
```
 
This is the direct implementation of what the theory describes: each layer transforms the representation, building more abstract features from the output of the previous one.
 
### Shapes flowing through the model
 
Tracking the shape of your tensor through each layer is one of the most useful debugging habits in PyTorch:
 
```python
x = torch.randn(1, 784)   # batch of 1 sample, 784 features
 
x = model[0](x)  # (1, 784) → (1, 128)
x = model[1](x)  # (1, 128) → (1, 128)  activation, shape unchanged
x = model[2](x)  # (1, 128) → (1, 64)
x = model[3](x)  # (1, 64)  → (1, 64)   activation, shape unchanged
x = model[4](x)  # (1, 64)  → (1, 10)
```
 
Activation functions (`nn.ReLU`) do not change the shape — they only transform the values. Only `nn.Linear` changes the dimensionality.
