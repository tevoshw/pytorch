# Dropout

A regularization technique that turns off some random neurons

`nn.Dropout(p=x)` X need to be between 0 and 1, 0.5 = 50% of the neurons will be deactivaded

The layer of dropout normally goes after the activation function

| | Expected format | Zeros | Use in |
|---|---|---|---|
| `Dropout` | `(batch, features)` | individual values | MLP, Linear |
| `Dropout1d` | `(batch, C, L)` | channel along time | NLP, audio |
| `Dropout2d` | `(batch, C, H, W)` | entire 2D map | CNN, images |
| `Dropout3d` | `(batch, C, D, H, W)` | entire 3D volume | video, MRI |

```python
import torch 
import torch.nn as nn


neurons = torch.rand(5, 10)
dropout = nn.Dropout(p=0.5)
output = dropout(neurons)

print(f'Neurons before:\n{neurons}\n')
print(f'Neurons after:\n{output}')


"""

Each column of the matrix it's a neuron, and in each rows (samples) the dropout turn off some random neurons

"""
```
