# Loss

## 1. Regression

1. `nn.MSELoss()`: MSE (L2 Loss) — $\frac{1}{n}\sum(\hat{y}_i - y_i)^2$
2. `nn.L1Loss()`: MAE (L1 Loss) — $\frac{1}{n}\sum|\hat{y}_i - y_i|$
3. `nn.HuberLoss(delta=1.0)`: Huber Loss — L2 when $|e| \leq \delta$, L1 when $|e| > \delta$. and $delta$ it's the hyperparameter that we can define
4. `torch.sqrt(nn.MSELoss()(pred, target))`: RMSE — square root of MSE, same unit as $y$


## 2. Classification

1. `nn.CrossEntropyLoss()`: The loss gonna log the probabilities of the output [0.7, 0.2, 0.1], will take the right position and log them, looks like = $-log(0.7)$
> The Cross Entropy in PyTorch Alredy have the softmax function inside, so do not use softmax in the architecture of the nn and use the ce!!!
2. `nn.BCEWithLogitsLoss()` or `nn.BCELoss()`: The CrossEntropy but for just 2 classes
> The main difference here it's `BCEWithLogitsLoss` alredy have the sigmoid function inside and `nn.BCELoss` don't, so do not use and put the af in the nn!!
3. `nn.NLLLoss()`: The Cross Entropy without the softmax applied, basically the cross entropy in pytorch are $softmax + Log_Class$ and the NLLLoss are the $Log_Class$



```python
import torch
import torch.nn.functional as F
import torch.nn as nn


# Regression
output = torch.randint(0, 10, (1, 10), dtype = torch.float32)
target = torch.randint(0, 10, (1, 10), dtype = torch.float32)

mse = nn.MSELoss()
mae = nn.L1Loss()
hubber = nn.HuberLoss(delta = 1)

loss_mse = mse(output, target)
loss_mae = mae(output, target)
loss_hubber = hubber(output, target)
rmse = torch.sqrt(mse(output, target))

print(f'=' *  50)
print(f'Regression Loss')
print(f'MSE: {loss_mse.item():.3f}, MAE (L1): {loss_mae.item():.3f}, Hubber: {loss_hubber.item():.3f}, RMSE: {rmse:.3f}')


# Classification
output = torch.randn(10, 100)       
target = torch.randint(0, 100, (10,))

output2 = torch.randint(0, 2, (1, 10), dtype = torch.float32)
target2 = torch.randint(0, 2, (1, 10), dtype = torch.float32)

bce = nn.BCEWithLogitsLoss()
cross = nn.CrossEntropyLoss()

loss_bce = bce(output2, target2)
loss_cross = cross(output, target)

print(f'=' *  50)
print(f'Classitication loss')
print(f'BCE: {loss_bce.item():.3f}, Cross: {loss_cross.item():.3f}')
```
