# 1. Dynamic PTQ Quantization
(torchao)


PerTensor, PerRow

> The inplace garanted that others copy of the model don't will make created, so ever that time that you use don't forget that the original wieghts gonna be losed

```python
import torch
import torch.nn as nn
from torchao.quantization import quantize_
from torchao.quantization import Int8DynamicActivationInt8WeightConfig, Int8StaticActivationInt8WeightConfig
from torchao.quantization import PerTensor

model = nn.Linear(32, 32).to(torch.bfloat16).cuda()

print(f'Model before the quantization weights')
for name, param in model.named_parameters():
    print(f'Name: {name}, Param: {param}')
print(f'=' * 50)

quantize_(
    model,
    Int8DynamicActivationInt8WeightConfig(
        granularity=PerTensor(),
        version=2,   
    ),
)

print(f'Model after quantization weights')
for name, param in model.named_parameters():
    print(f'Name: {name}, Param: {param}')
print(f'=' * 50)
```
