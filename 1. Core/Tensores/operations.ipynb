# 1. Mul (Element-wise Multiplication or *)

Multiplies corresponding elements. Output shape equals input shape

For this operations occurs both vector, matrixs or tensors needs to have the exactly same shape

- We can use two options:
1. PyTorch: `torch.mul(x, y)`
2. Python : `x * y`

````python
matrix1 = torch.rand(2, 2)
matrix2 = torch.rand(2, 2)

# Way1
mul1 = torch.mul(matrix1, matrix2)
# Way 2
mul2 = matrix1 * matrix2

print(f'Matrix 1 Elements:\n{matrix1}')
print(f'Matrix 2 Elements:\n{matrix2}')

print(f'Mult 1:\n{mul1}')
print(f'Mult 2:\n{mul2}')

"""

0.1992 * 0.2706, 0.1971 * 0.1751
0.3914 * 1030, 0.1575 * 0.1166


"""
````

# 2. Dot Product 

A dot product multiplies corresponding elements of two vectors and sums the results into a single scalar. It is the fundamental operation that measures **how much two vectors align**

For this operations the shapes need to be x = (m, n) // y = (n, p), and the output shape gonna be (m, p)


- We same some options:
1. (1D): `torch.dot(x, y)`
2. (2D+): `torch.matmul(x, y)`
3. (2D+): `x @ y`

````python

matrix1 = torch.rand(5)
matrix2 = torch.rand(5)
matrix3 = torch.rand(2, 3)
matrix4 = torch.rand(3, 2)

# Way1
mul1 = torch.dot(matrix1, matrix2)

# Way 2
mul2 = torch.matmul(matrix3, matrix4)

# Way 3
mul3 = matrix3 @ matrix4

print(f'Matrix 1 Elements:{matrix1}')
print(f'Matrix 2 Elements:{matrix2}')

print(f'Mult 1 = {mul1:.3f}\n')
print(f'=' * 50)

print(f'Matrix 3 Elements:\n{matrix3}\n')
print(f'Matrix 4 Elements:\n{matrix4}\n')
print(f'Mult 2:\n{mul2}\n')
print(f'Mult 3:\n{mul3}\n')
````

# 3. Unary


1. Abs: Removes the sign — negative values turn positive, positive become equal.
2. Neg: Inverts the sign of each element.
3. Sqrt: Square root. It only works at negative values ≥ 0 — return nan.
4. Exp: Raises e ≈ 2,718 to the power of each element.
5. Log: Natural logarithm. Inverse of exp. Only for values > 0 — zero and negative return -inf or nan.
6. Clamp: Limits values between a minimum and a maximum. Values outside the range are cut
7. Sign: Returns only the sign: -1, 0, or +1.



````python
one = torch.tensor([-3])
two = torch.tensor([144])
tree = torch.tensor([2])

abs = torch.abs(one)
neg = torch.neg(one)
sqrt = torch.sqrt(two)
exp = torch.exp(tree)
log = torch.log10(tree) # basis 10
clamp = torch.clamp(two, max=10, min=1)
sign = torch.sign(tree)

print(f'ABS of {one[0]} = {abs}')
print(f'NEG of {one[0]} = {neg}')
print(f'SQRT of {two[0]} = {sqrt}')
print(f'EXP of {tree[0]} = {exp}')
print(f'LOG of {tree[0]} = {log}')
print(f'CLAMP of {two[0]} (min 1, max 10) = {clamp}')
print(f'SIGN of {tree[0]} = {sign}')
````

# 4. Broadcasting — NumPy / PyTorch

Lets you operate on arrays/tensors of **different shapes** without manually resizing them. The smaller array is automatically "stretched" to match the bigger one.

## Scalar + Array

```python
arr = np.array([1, 2, 3])
arr + 10
# -> [11, 12, 13]
```

`10` is stretched to `[10, 10, 10]` automatically.

## Matrix + Vector

```python
mat = np.array([[1,2,3],
                  [4,5,6]])
vec = np.array([10,20,30])

mat + vec
# -> [[11,22,33],
#     [14,25,36]]
```

`vec` (shape `(3,)`) is applied to **every row** of `mat` (shape `(2,3)`).

## Broadcasting Rule

Two dimensions are compatible if:
- they are equal, OR
- one of them is 1 (gets stretched)

```python
mat.shape   # (2, 3)
vec.shape   # (3,)  -> treated as (1,3), stretched to (2,3)
```

## Common ML Use

```python
X = np.random.rand(100, 5)   # 100 samples, 5 features
mean = X.mean(axis=0)          # shape (5,)
X_norm = X - mean                 # broadcasting applies subtraction per row
```

Used constantly for: normalization, adding bias per neuron, combining embeddings — no loops needed.

# 5. Reductions — NumPy / PyTorch

Operations that collapse an array/tensor into fewer values (sum, mean, max, etc).

## No axis — reduces everything to one value

```python
mat = np.array([[1,2,3],
                  [4,5,6]])

mat.sum()    # 21
mat.mean()   # 3.5
mat.max()    # 6
```

## With `axis` (NumPy) / `dim` (PyTorch) — reduces only one dimension

```python
mat.sum(axis=0)   # sum per COLUMN -> [5, 7, 9]
mat.sum(axis=1)   # sum per ROW    -> [6, 15]
```

**How to think about `axis`:** it's the axis that **disappears** after the operation.

- `axis=0` → collapses rows → 1 value per **column**
- `axis=1` → collapses columns → 1 value per **row**

```python
mat.shape                # (2, 3)
mat.sum(axis=0).shape    # (3,)  -> rows gone
mat.sum(axis=1).shape    # (2,)  -> columns gone
```

## Common reductions

```python
arr.sum(axis=0)      # sum
arr.mean(axis=0)     # mean
arr.max(axis=0)      # max value
arr.min(axis=0)      # min value
arr.std(axis=0)      # standard deviation
arr.var(axis=0)      # variance
arr.argmax(axis=0)   # INDEX of the max (not the value)
arr.argmin(axis=0)   # index of the min
arr.prod(axis=0)     # product of all values
arr.any(axis=0)      # is any True?
arr.all(axis=0)      # are all True?
```

## Common ML use

```python
X.mean(axis=0)   # per-feature mean (used for normalization)
X.std(axis=0)    # per-feature std

# classic example: get predicted class
probs = torch.tensor([[0.1, 0.7, 0.2],
                        [0.6, 0.3, 0.1]])
probs.argmax(dim=1)   # -> [1, 0] (index of highest prob per row)
```
