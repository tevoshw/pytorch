## 1. Why PyTorch?

Many people are familiar with NumPy, so why use PyTorch instead?

One of the main reasons is **GPU acceleration**, which allows us to perform computationally intensive operations much faster by leveraging parallel processing.

Although NumPy is highly efficient for numerical computations, it primarily operates on the CPU. PyTorch, on the other hand, supports both CPU and GPU computations, making it particularly useful for Machine Learning and Deep Learning, where large amounts of data and complex mathematical operations are involved.

Additionally, PyTorch provides **automatic differentiation** through its autograd engine, allowing us to compute gradients automatically during model training.

## 2. Creating Tensors

As in NumPy, PyTorch offers several ways to create tensors (a matrix is just a 2D tensor). Let's look at the most common ones.

### From existing data

- `torch.tensor([x, y, ..., z])`: creates a tensor from the given values (lists, nested lists, or NumPy arrays).

### Tensors with fixed values

- `torch.zeros(x, y)`: tensor of shape `(x, y)` filled with `0`.
- `torch.ones(x, y)`: tensor of shape `(x, y)` filled with `1`.
- `torch.full((x, y), v)`: tensor of shape `(x, y)` filled with the value `v`.
- `torch.eye(n)`: `n x n` identity matrix.
- `torch.zeros_like(t)`: zeros with the same shape and dtype as tensor `t` (also `ones_like`, `full_like`).

### Sequences and ranges

- `torch.arange(start, end, step)`: values from `start` up to, but **not including**, `end`, spaced by `step`.
- `torch.linspace(start, end, steps=n)`: exactly `n` evenly spaced values from `start` to `end`, **both included**.



### Random tensors

- `torch.rand(x, y)`: uniform random values in `[0, 1)`.
- `torch.randn(x, y)`: random values from a standard normal distribution (mean `0`, standard deviation `1`).
- `torch.randint(a, b, (x, y))`: random integers from `a` up to, but **not including**, `b`.
> `torch.manual_seed(42)`: call it before generating random tensors to make results reproducible.

### Useful notes

- Every function accepts `dtype=` and `device=`, e.g. `torch.zeros(3, 3, dtype=torch.float32, device="cuda")`.
- The default float type is `float32` (NumPy uses `float64`).
- Check a tensor with `t.shape`, `t.dtype` and `t.device`.

```python
import torch

zeros_m = torch.zeros(2,2)
ones_m = torch.ones(2, 2)
fulls_m = torch.full((2, 2), 4)
eye_m = torch.eye(4)
like_z_m = torch.zeros_like(zeros_m)


print(f'Matrix of zeros:\n{zeros_m}\n')
print(f'Matrix of ones:\n{ones_m}\n')
print(f'Matrix of fulls:\n{fulls_m}\n')
print(f'Eye matrix:\n{eye_m}\n')
print(f'Matrix like zeros:\n{like_z_m}\n')

arange = torch.arange(0, 10, 2)
linspace = torch.linspace(0, 10, 5)

print(f'Matrix of arange: {arange}')
print(f'Matrix of linspace: {linspace}\n')


random = torch.rand(2, 2)
randomn = torch.randn(2, 2)
randint = torch.randint(0, 10, (2, 2))

print(f'Matrix of random: \n{random}\n')
print(f'Matrix of randomn: \n{randomn}\n')
print(f'Print of random int nunbers: \n{randint}\n')
```

### 2.1 `torch.tensor` vs `torch.Tensor`

1. `torch.tensor` Create a tensor and you can change the dtype
2. `torch.Tensor` Create a tensor and you can't change the dtype, it always float32

```python
tensor1 = torch.tensor([1,2,3])
tensor2 = torch.Tensor([1,2,3])

print(f'Tensor1:\n{tensor1}, dtype: {tensor1.dtype}\n')
print(f'Tensor2:\n{tensor2}, dtype: {tensor2.dtype}')
```
