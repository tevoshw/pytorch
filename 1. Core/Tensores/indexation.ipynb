# Indexation

Techniques to acess data inside the arrays/matrixs/tensors. When you create an array will take up a space in the memory. So when you do a indexation operation we can create another array with the values or just have acess of this space in the memory.

# View vs Copy — NumPy / PyTorch
 
| Operation | View or Copy? |
|---|---|
| Basic slicing (indexing) `arr[1:3]`, `arr[::2]`, `mat[:, 1]` | **View** |
| Fancy indexing `arr[[1,3,5]]` | **Copy** |
| Boolean indexing `arr[arr > 10]` | **Copy** |



```python
import torch
```

### 1. Basic Indexation (per position)

We gonna get an element or a part (slice) using the tensor name with [] and the position that we want `tensor[position]` for 1 element. For more we can slice using the logic `start:stop:step` inside the [], so `tensor[0:2:1]`.

> Always gonna be the view, so don't create other object in the memory

```python
tensor = torch.arange(11)
print(f'Tensor created: {tensor} and we want to acess the second element (1), so: {tensor[1]}')
print(f'Tensor created: {tensor} and we want to acess the all the elements between the second and the seventh in a step of 1, so: {tensor[1:7:1]}')


""" 
More exemples below
"""
```

```python
tensor[::2]
```

```python
tensor[-3:]
```

```python
tensor[:-2:2]
```

```python
tensor[:]
```

### 2. Fancy Indexation

Gonna get the elements by a specific list. Always a array inside the [], so `tensor[ [] ]`

> Always returns a copy

```python
idx = [2,3,4]

tensor2 = torch.arange(11)
print(tensor2[idx])

"""
1D TECHNIQUE: Get the idx 
"""
```

```python
tensor2 = torch.randn(4,4)
print(f'Complete matrix:\n{tensor2}')
print(f'Selected rows of the matrix (2 and 3):\n{tensor2[[1,2]]}')

"""
2D TECHNIQUE 1: Get the row complete
"""
```

```python
print(f'Complete matrix:\n{tensor2}')
print(f'Selected points of the matrix (1,1) and (2,2):\n{tensor2[[1, 2], [1, 2]]}')

"""
2D TECHNIQUE 1: Get the specific points (x, y)
# Logic [row], [col] -> so [1, 2], [2, 2]

Means get the row idx (1) with the col idx 2 (1, 2)
"""
```

### 3. Boolean Indexation

Select elements using conditions

> Always returns a copy

```python
tensor3 = torch.arange(10)
print(f'Selecting elemnts bigger than 5: {tensor[tensor > 5]}')
```

### 4. Gather

"Give me the value of this idx"

```python
tensor4 = torch.rand(4, 4)
idx = torch.tensor([0, 1, 2, 3])

print(f'Matrix complete:{tensor4}')
print(tensor4.gather(1, idx.unsqueeze(1)))
```

### 5. Scatter

Some logic of gather but instead of read, gonna write a value

```python
tensor4 = torch.rand(4, 4)
idx = torch.tensor([0, 1, 2, 3])


print(f'Matrix before:\n{tensor4}')

print()
print(f'Tensor after the scatter:\n{tensor4.scatter(1, idx.unsqueeze(1), 10)}')
```
