## Det

The **determinant** is a scalar computed from a square matrix. It tells you whether the matrix is invertible and how it scales volume.

- `det(A) != 0` → A is invertible
- `det(A) == 0` → A is singular (no inverse)
- `|det(A)|` = factor by which A scales area (2D) or volume (3D)
- The sign indicates whether orientation is flipped

```python
import torch

A = torch.tensor([[2., 1.],
                  [1., 3.]])

torch.linalg.det(A)            # tensor(5.)   -> 2*3 - 1*1
```

For 2x2 matrices: `det([[a, b], [c, d]]) = ad - bc`

For large matrices, use `slogdet` to avoid overflow/underflow:

```python
sign, logabsdet = torch.linalg.slogdet(A)
```

Useful properties:

- `det(AB) = det(A) * det(B)`
- `det(A^T) = det(A)`
- `det(A^-1) = 1 / det(A)`
- `det(A)` = product of the eigenvalues


````python
import torch

matrix = torch.rand(2, 2)
det = torch.linalg.det(matrix)
print(f'Matrix: {matrix}')
print(f'Det: {det}')
````


## Inverse

The **inverse** of A is the matrix A⁻¹ such that:

`A @ A⁻¹ = A⁻¹ @ A = I`

It only exists for square matrices with `det(A) != 0`.

```python
A = torch.tensor([[2., 1.],
                  [1., 3.]])

A_inv = torch.linalg.inv(A)

# Check: should be the identity (up to floating-point error)
torch.allclose(A @ A_inv, torch.eye(2), atol=1e-6)   # True
```



````python
inverse = torch.linalg.inv(matrix)
print(f'Matrix: {matrix}')
print(f'Inverse: {inverse}')
print(f'Checking: {torch.allclose(torch.eye(2), matrix @ inverse, atol=1e-6)}')
````


## Eye

`torch.eye(n)` creates the **identity matrix**: ones on the main diagonal, zeros elsewhere. It is the neutral element of matrix multiplication: `A @ I = I @ A = A`.

```python
torch.eye(3)
# tensor([[1., 0., 0.],
#         [0., 1., 0.],
#         [0., 0., 1.]])

torch.eye(2, 3)                    # non-square: 2 rows, 3 columns
torch.eye(3, dtype=torch.float64)  # choose the dtype
```

Common uses:

- Checking an inverse: `A @ A_inv ≈ torch.eye(n)`
- Regularization: `A + lambda * torch.eye(n)` (ridge regression, keeps matrices invertible)
- Initializing weights or transformations as "do nothing"
- Batch of identities: `torch.eye(3).expand(32, 3, 3)`
