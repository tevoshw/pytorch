## Solving a Linear System
Given the linear system

$Ax = b$

where:
- $(A)$ is the coefficient matrix.
- $(x)$ is the vector of unknowns.
- $(b)$ is the vector of known values.

If $A$ is square and invertible, we can isolate $x$ by multiplying both sides by the inverse of $A$:

 $[A^{-1}Ax = A^{-1}b]$

Since $(A^{-1}A = I)$, and $(Ix = x)$, we obtain:

$[{x = A^{-1}b}]$

Using the function `np.linalg.solve(A, b)` solves the system of linear equations directly. We could manually calculate the inverse of `A` and multiply it by `b`, but using `linalg.solve` is generally more efficient and numerically stable.

```python

torch.linalg.solve(A, b)

```



````python
import torch

A = torch.tensor([[2., 1.],
                  [1., 3.]])
b = torch.tensor([5., 10.])

result = torch.linalg.solve(A, b) # Better 
print(result)
````
