# Linear Algebra in PyTorch

````python
import torch

A = torch.tensor([1,2,3])
B = torch.tensor([10,20,30])
````

# Dot Product

The **dot product** (scalar product) takes two vectors of the same size and returns a **scalar**:

`a · b = Σ aᵢ bᵢ = ‖a‖ ‖b‖ cos(θ)`

It measures how much two vectors point in the same direction.

```python
import torch

a = torch.tensor([1., 2., 3.])
b = torch.tensor([4., 5., 6.])

torch.dot(a, b)        # tensor(32.)  -> 1*4 + 2*5 + 3*6
a @ b                  # same result
```

Properties:

- Commutative: `a · b = b · a`
- `a · a = ‖a‖²`
- `a · b = 0` → vectors are orthogonal
- Positive if the angle is < 90°, negative if > 90°

Notes:

- `torch.dot` only works with 1D tensors (no batch).
- For batches, use `torch.linalg.vecdot(A, B, dim=-1)` or `torch.einsum('bi,bi->b', A, B)`.
- Used in ML for cosine similarity, linear neurons (`w · x + b`), and attention scores (`Q @ K.mT`).


````python
dot = torch.dot(A, B)
print(dot)
print(dot.shape)
````


# Cross Product

The **cross product** (vector product) takes two **3D** vectors and returns a **vector** perpendicular to both:

`a × b = (a₂b₃ − a₃b₂, a₃b₁ − a₁b₃, a₁b₂ − a₂b₁)`

Its magnitude is `‖a‖ ‖b‖ sin(θ)`, which equals the **area of the parallelogram** spanned by a and b. The direction follows the right-hand rule.

```python
a = torch.tensor([1., 2., 3.])
b = torch.tensor([4., 5., 6.])

torch.linalg.cross(a, b)   # tensor([-3., 6., -3.])
```

Properties:

- Anti-commutative: `a × b = −(b × a)`
- `a × b = 0` → vectors are parallel
- Result is orthogonal to both inputs: `(a × b) · a = 0`
- Not associative

Notes:

- The dimension used must have size **3**, otherwise PyTorch raises an error.
- Use `torch.linalg.cross` (default `dim=-1`) instead of the legacy `torch.cross`.
- Common uses: surface normals, triangle area, torque, building orthonormal bases.
- Scalar triple product: `a · (b × c)` gives the volume of the parallelepiped.


````python
vetor = torch.linalg.cross(A, B)
print(vetor)
````


# Outer Product

The **outer product** takes two vectors of any sizes (m and n) and returns an **m×n matrix**:

`(a ⊗ b)ᵢⱼ = aᵢ bⱼ`, equivalent to `a bᵀ`

```python
a = torch.tensor([1., 2., 3.])
b = torch.tensor([4., 5.])

torch.outer(a, b)
# tensor([[ 4.,  5.],
#         [ 8., 10.],
#         [12., 15.]])

a[:, None] * b[None, :]           # same result (broadcast)
torch.einsum('i,j->ij', a, b)     # same result
```

Properties:

- The result always has **rank 1** (every row is a multiple of b)
- `tr(a bᵀ) = a · b`
- `‖a bᵀ‖_F = ‖a‖ ‖b‖`
- `(a bᵀ) v = a (b · v)`

Notes:

- `torch.outer` only accepts 1D tensors. For batches, use `torch.einsum('bi,bj->bij', A, B)`.
- For higher-order tensors, use `torch.tensordot(A, B, dims=0)`.
- Common uses: gradient of a linear layer (`dL/dW = outer(δ, x)`), covariance matrices, projectors (`u uᵀ`), and low-rank methods such as LoRA.

````python
outer = torch.outer(A, B)
print(outer)
````
