```python
import torch

m = torch.rand(10, 12)
print(f'Shape of original matrix: {m.shape}')
```

# Tensor Shape

## Reshaping

- **`.view(*shape)`**
  - Returns a new tensor that **shares memory** with the original.
  - Requires a compatible memory layout (usually contiguous); otherwise it raises a `RuntimeError`.
  - Never copies data.

- **`.reshape(*shape)`**
  - Returns a view when possible; otherwise it **copies** the data.
  - Works on non-contiguous tensors.
  - Safer, but you can't always tell whether the result shares memory.

## Quick Comparison

| Method     | Copies data?      | Needs contiguous? | Shares memory? |
|------------|-------------------|-------------------|----------------|
| `.view`    | Never             | Yes               | Always         |
| `.reshape` | Only if necessary | No                | When possible  |


> -1 is a way to let PyTorch automatically calculate a dimension based on the total number of elements and the other dimensions reported.

```python
reshape_m = m.reshape(6, 20)
view_m = m.view(20, 6)


print(f'Shape after the transpose: {reshape_m.shape}')
print(f'Shape after the permute: {view_m.shape}')
```

# Transpose

## Swapping and Reordering Dimensions

- **`.transpose(dim0, dim1)`**
  - Swaps exactly **two** dimensions.
  - Returns a view that **shares memory** with the original (no copy).
  - The result is usually **non-contiguous**.

- **`.permute(*dims)`**
  - Reorders **all** dimensions in one call, given the new order as a tuple of indices.
  - Returns a view that **shares memory** with the original (no copy).
  - The result is usually **non-contiguous**.

## Quick Comparison

| Method       | Dims affected   | Arguments            | Copies data? | Shares memory? |
|--------------|-----------------|----------------------|--------------|----------------|
| `.transpose` | Two             | `(dim0, dim1)`       | Never        | Always         |
| `.permute`   | Any / all       | Full new dim order   | Never        | Always         |




```python
transpose_m = m.transpose(1, 0)
permute_m = m.permute(1, 0)


print(f'Shape after the transpose: {transpose_m.shape}')
print(f'Shape after the permute: {permute_m.shape}')
```

# Squeeze

## Removing and Adding Singleton Dimensions

- **`.squeeze(dim=None)`**
  - Removes dimensions of size **1**.
  - Without `dim`, removes **all** size-1 dimensions; with `dim`, removes only that one (and does nothing if its size isn't 1).
  - Returns a view that **shares memory** with the original.

- **`.unsqueeze(dim)`**
  - Inserts a new dimension of size **1** at position `dim`.
  - Accepts negative indices (e.g. `-1` appends a dimension at the end).
  - Returns a view that **shares memory** with the original.

## Quick Comparison

| Method       | Effect on shape        | Arguments                | Copies data? | Shares memory? |
|--------------|------------------------|--------------------------|--------------|----------------|
| `.squeeze`   | Removes size-1 dims    | Optional `dim` (or tuple)| Never        | Always         |
| `.unsqueeze` | Adds one size-1 dim    | Required `dim`           | Never        | Always         |


```python
m = torch.rand(1, 10, 10)

squeeze_m = m.squeeze(0)
unsqueeze_m = m.unsqueeze(3)

print(f'Original matrix shape: {m.shape}')
print(f'Shape after the transpose: {squeeze_m.shape}')
print(f'Shape after the permute: {unsqueeze_m.shape}')
```

# Concat

## Joining Tensors

- **`torch.cat([tensors], dim=0)`**
  - Concatenates tensors along an **existing** dimension.
  - All tensors must have the **same shape** except along `dim`.
  - The number of dimensions stays the same; the size along `dim` is the sum of the inputs.
  - Always **copies** data into a new tensor.

- **`torch.stack([tensors], dim=0)`**
  - Joins tensors along a **new** dimension inserted at `dim`.
  - All tensors must have **exactly the same shape**.
  - The result has one more dimension than the inputs.
  - Always **copies** data into a new tensor.

## Quick Comparison

| Method        | Dimension used | Shape requirement           | Output ndim | Copies data? |
|---------------|----------------|-----------------------------|-------------|--------------|
| `torch.cat`   | Existing       | Equal except along `dim`    | Same        | Always       |
| `torch.stack` | New            | Identical                   | +1          | Always       |

```python
m1 = torch.rand(5, 5)
m2 = torch.rand(5, 5)

cat_m = torch.cat([m1, m2], dim = 1)
stack_m = torch.stack([m1, m2])

print(f'Original matrix 1 shape: {m1.shape}')
print(f'Original matrix 2 shape: {m2.shape}')
print(f'Shape after the cat in dim 1: {cat_m.shape}')
print(f'Shape after the stack: {stack_m.shape}')
```

# Flatten

## Collapsing Dimensions

- **`.flatten(x, start_dim=0, end_dim=-1)`**
  - Collapses the dimensions from `start_dim` to `end_dim` (inclusive) into a **single** dimension.
  - With no arguments, flattens the whole tensor into 1D.
  - Returns a **view** when possible (contiguous layout); otherwise it **copies**, like `.reshape`.
  - A 0-dim tensor becomes a 1D tensor of size 1.


```python
flatten_m  = torch.flatten(m, start_dim = 1)
print(f'Original matrix: {m.shape}')
print(f'Matrix after the flatten: {flatten_m.shape}')
```
